The winning beaches from the Keep Britain Tidy webpage have been stored in a file. Each row of the file corresponds to a beach, and stores the name of the beach, the council responsible for the beach, and the region of the UK.

Furthermore each council has a beach manager. Beach managers receive a bonus of £100 for every Blue Flag beach that they oversee, with a further £200 awarded for every Blue Flag beach over the national council average.

You have been asked to produce Python code that can be used now and in future years to calculate the number of Blue Flag beaches per council, the number of Blue Flag beaches per region, and bonuses for beach managers. You should communicate the 2025 figures explicitly and analyse the mathematical properties of the allocation of bonuses.
