## What is Data Ingress?
Data ingress means getting data into a program. Data can come from files, databases, websites, sensors, and other sources. It is the first step in a data science workflow.

## 2.1 Input and Output (I/O)
Input = data coming into a program.

Output = data leaving a program.

Examples: reading files, displaying results, saving reports.

```python
import random

MAX_ATTEMPTS = 3
secret = str(random.randint(1, 10))

for attempt in range(MAX_ATTEMPTS):
    guess = input("Please enter a number: ")
    if guess == secret:
        print("Congratulations! You guessed the number.")
        break
    print("Not quite!")
else:
    print(f"Out of attempts. I was thinking of {secret}")
```

### 2.1.2 File Paths
Relative paths are preferred because they work on different computers.

Example relative path: `data/steps.csv` means “the file properties.txt, inside the folder data, inside the working directory”

### 2.1.3 Reading Files

When we open a file, we get back a file object, with methods for reading and writing, such as `read`, `readline`, and `write`. When we’re done with the file, we must close it with the `close` method, to free up the resources the operating system has given us

The `with` statement creates a context manager, which automatically closes the file when the indented block finishes, even if an error occurs. This is the recommended way to work with files in Python, and you should use it whenever you call `open`.

```python
filename = "data/properties.txt"

# open the file for reading using a context manager
with open(filename, "r") as file:
    # read the whole contents of the file into a single string
    contents = file.read()

# the file is automatically closed when we exit the with block
print(contents)
```

Python provides several ways to read files. 
- `read()` returns the entire file as one string
- `readlines()` returns a list containing each line
- `readline()` reads a single line at a time. 

Lines read from a file usually include a newline character (`\n`), which can be removed using the `strip` method. A common and efficient approach is to loop directly over the file object, processing one line at a time instead of loading the whole file into memory, making it more suitable for large files.

```python
addresses = []
with open("data/properties.txt", "r") as file:
    for line in file:  # loop over the file one line at a time
        addresses.append(line.strip())

print(addresses)
```

### 2.1.4 Writing Files

Writing to a file in Python is similar to reading from one, but files are opened in write ("w") or append ("a") mode and data is written using the `write` method. Unlike `print`, `write` does not automatically add a newline, so `\n` must be included manually. Files opened in write mode are created automatically if they do not already exist. Python also allows multiple files to be opened in a single `with` statement, making it easy to read from one file, process the contents, and write the results to another file.

```python
input_filename = "data/properties.txt"
output_filename = "data/properties_upper.txt"

# open the input file for reading and the output file for writing
with (
    open(input_filename, "r") as input_file,
    open(output_filename, "w") as output_file,
):
    for line in input_file:
        # process the line (in this case, convert to upper case)
        processed_line = line.strip().upper()
        # write the processed line to the output file
        output_file.write(processed_line + "\n")
        # equivalently: print(processed_line, file=output_file)
```

## 2.2 CSV Files
CSV stands for Comma-Separated Values.

Each line in the file represents a row of data, and each value in the row is separated by a comma. The first line of the file usually contains the names of the columns, which is known as the header row. 

### 2.2.1 Reading CSV Files

Python’s built-in csv module provides a reliable way to read CSV files because simply splitting lines on commas can fail when data values contain commas.
A CSV reader returns each row as a list, and the header row is usually handled separately using `next()`. 

All values are initially read as strings, so the programmer must convert them to the appropriate data types, such as `int` or `float`, before performing calculations or comparisons. Unpacking can also be used to assign each value in a row to a named variable, making the code easier to read and work with.

```python
import csv
with open("data/steps.csv", "r") as f:
    reader = csv.reader(f)
    header = next(reader)
    print(header)
    for row in reader:
        print(row)
```

```python
with open("data/steps.csv", "r") as steps_file:
    reader = csv.reader(steps_file)
    header = next(reader)

    for row in reader:
        # unpack the row elements into three variables
        step, rise, going = row

        # convert each value to the appropriate type
        step = int(step)
        rise = float(rise)
        going = float(going)

        print([step, rise, going])
```

A `csv.writer` object writes rows to the output file, while the header row is modified separately to include the new column name (`"valid"`). 

Each row is read, converted to the correct data types, validated, and then written to the new file along with its validation result. 

When writing CSV files, `newline=""` should be included to avoid extra blank lines on Windows.

```python
import csv


def valid_rise(rise):
    """Validate the rise value."""
    return False  # placeholder until we write the real check


def valid_going(going):
    """Validate the going value."""
    return False  # placeholder until we write the real check


with (
    open("data/steps.csv", "r") as steps_file,
    open("data/steps_validated.csv", "w", newline="") as validated_file,
):
    reader = csv.reader(steps_file)
    writer = csv.writer(validated_file)

    # add a new column to the header for validation, and write it out
    header = next(reader)
    header.append("valid")
    writer.writerow(header)

    for row in reader:
        step, rise, going = row
        step = int(step)
        rise = float(rise)
        going = float(going)

        # validate the rise and going values
        valid = valid_rise(rise) and valid_going(going)

        # write the row, with its validation result, to the output file
        writer.writerow([step, rise, going, valid])
```

#### Validating Data

The final step is to implement the validation logic. The functions `valid_rise` and `valid_going` should return True if the rise and going values are within the regulations, and False otherwise. 

Note that `a <= b <= c` is a common Python idiom for checking if b is between a and c, inclusive. This is equivalent to `a <= b and b <= c`, but is arguably slightly easier to read.

```python
import csv


def valid_rise(rise):
    """Validate the rise value."""
    return 150 <= rise <= 220


def valid_going(going):
    """Validate the going value."""
    return 220 <= going <= 300


with (
    open("data/steps.csv", "r") as steps_file,
    open("data/steps_validated.csv", "w", newline="") as validated_file,
):
    reader = csv.reader(steps_file)
    writer = csv.writer(validated_file)

    # add a new column to the header for validation, and write it out
    header = next(reader)
    header.append("valid")
    writer.writerow(header)

    for row in reader:
        step, rise, going = row
        step = int(step)
        rise = float(rise)
        going = float(going)

        # validate the rise and going values
        valid = valid_rise(rise) and valid_going(going)

        # write the row, with its validation result, to the output file
        writer.writerow([step, rise, going, valid])
```

### 2.2.3 Key insights to remember when reading and writing CSV files
1. Use the csv module, not manual string splitting
    - CSV files should be read and written using Python's csv module because it correctly handles commas, quotes, and other CSV formatting rules.
    - Avoid using split(",") for real CSV data.
2. Treat the header separately
    - Use next(reader) to read the header row.
    - This allows you to inspect, modify, or extend column names before processing the data.
3. CSV data is read as strings
    - The csv module does not automatically convert data types.
    - Convert values to int, float, or other appropriate types before performing calculations or comparisons.
4. Use csv.writer to create output files
    - Create a writer object with csv.writer().
    - Use writerow() to write one row at a time as a list.
    - `writer.writerow([step, rise, going, valid])`
5. Add new columns by modifying the header
    - If creating derived data (such as validation results), add a new column name to the header and write it before the data rows.
    - `header.append("valid")`
    - `writer.writerow(header)`
6. Build a program skeleton first
    - Create placeholder functions and file-writing logic before implementing the full processing rules.
    - This helps test the workflow and output format early.
7. Use newline="" when writing CSV files
    - Prevents extra blank lines appearing in CSV files on Windows.
    - `open("output.csv", "w", newline="")`
8. Avoid magic numbers
    - Store important values such as limits in named constants rather than hard-coding them.
    - `DWELLING_RISE_MM = (150, 220)`
    - `DWELLING_GOING_MM = (220, 300)`
9. Write clear documentation
    - Function docstrings should explain:
        - What the function does
        - Expected input types
        - Valid ranges or rules
        - What the function returns
10. Make validation functions reusable
    - Pass limits as parameters instead of embedding them inside functions.
    - This makes the same code work for different staircase categories or future requirements.
    - `def valid_rise(rise, limits=DWELLING_RISE_MM):`
    - `minimum, maximum = limits`
    - `return minimum <= rise <= maximum`
11. Validate the whole dataset when necessary
    - Not all checks can be performed row by row.
    - Some validations require comparing rows or checking whether data is missing.
12. Do not assume row order
    - CSV rows may not be ordered correctly.
    - Use identifiers (such as step numbers) rather than relying on the order of rows in the file.
13. Handle missing data explicitly
    - If required information is missing (for example, step 1), report that there is insufficient information rather than incorrectly marking data as valid or invalid.

## 2.3 JSON Files

Very common for web APIs and configuration files.

Unlike CSV files, which are limited to tabular data, JSON files can represent much more complex, nested data structures.

JSON uses two main structures: objects and arrays.

Objects are the same as Python dictionaries. A JSON object is a collection of key-value pairs, where the keys are strings and the values can be any valid JSON data type: a string, number, Boolean , array, object, or null. Objects are enclosed in curly braces ({}), and the key-value pairs are separated by commas. For example:


    "name": "University of Nottingham",
    "founded": 1881,
    "active": true


Arrays are the same as Python lists. A JSON array is an ordered list of values, which can be of any valid JSON data type. Arrays are enclosed in square brackets ([]), and, like objects, the values are separated by commas. For example:

    "apple",
    7.5,
    false


The format is strict: keys must be enclosed in double quotes, single quotes are not valid, and trailing commas (a comma after the last item in an object or array) are not allowed, even though Python happily accepts them in dictionaries and lists. JSON also doesn’t support comments.

The great thing about JSON is that it can represent complex, nested data structures. For example, returning to our staircases, we could represent a list of housing properties, each with its own staircase specification, in a file called properties.json:

    
        "property_id": 1,
        "address": "123 Main St",
        "staircase": 
            "steps": 10,
            "rise": 180,
            "going": 250
        
    ,
    
        "property_id": 2,
        "address": "456 Elm St",
        "staircase": 
            "steps": 12,
            "rise": 200,
            "going": 300
        
    


Python’s built-in json module converts JSON into the corresponding Python data structures, and back again. The json.load function reads JSON from a file, giving us dictionaries and lists, and json.dump writes a Python object to a file as JSON.

We access the nested data by chaining indices and keys together from left to right: `house["staircase"]["steps"]` looks up the `"staircase" `dictionary within `house`, then the `"steps"` value within that. Note that we’ve avoided calling the loop variable `property`, as that’s the name of one of Python’s built-in functions

```python
import json

with open("data/properties.json", "r", encoding="utf-8") as json_file:
    # load the json data into regular python data structures
    properties = json.load(json_file)

# the outer structure is a list of length two
print(f"Data is a {type(properties)} of length {len(properties)}.")

# the first element is a dictionary with an "address" key
print(f"First address: {properties[0]['address']}")

# as it's just python data structures, we can process it as normal
for house in properties:
    address = house["address"]
    steps = house["staircase"]["steps"]
    print(f"Address {address} has {steps} steps.")
```

Let’s also look at writing JSON data to a file with json.dump. The indent argument makes the output much more readable, by putting each value on its own line and indenting nested structures.

There are also json.loads and json.dumps (with an “s”, for “string”), which convert between JSON stored in a string and Python data structures, rather than working with files.

```python
import json

# python data structure
data = [
    {
        "property_id": 1,
        "address": "123 Main St",
        "staircase": {"steps": 10, "rise": 180, "going": 250},
    },
    {
        "property_id": 2,
        "address": "456 Elm St",
        "staircase": {"steps": 12, "rise": 200, "going": 300},
    },
]

with open("data/output_properties.json", "w", encoding="utf-8") as json_file:
    # write the data to the file as json
    json.dump(data, json_file, indent=2)
```

## 2.4 XML Files
XML (eXtensible Markup Language) is a markup language used to store and transfer structured data in a format that is machine-readable and somewhat human-readable. It is commonly used for web services, configuration files, and formats such as SVG (Scalable Vector Graphics). Unlike raster images (JPEG/PNG), SVG uses geometric shapes and can be scaled without losing quality.

XML documents usually follow a schema, which defines the rules and structure that the document must follow. This makes XML more rigid but also more precise than JSON. XML data is organised using elements (defined by opening and closing tags), attributes (extra information inside tags), and nested structures that create a hierarchy of data. Some elements are self-closing, meaning they do not require a separate closing tag. HTML, the language used for web pages, is closely related to XML and uses the same concept of nested tagged elements.

```python
from IPython.display import SVG, display

svg = """<svg width="5cm" height="5cm" viewBox="0 0 200 200"
    xmlns="http://www.w3.org/2000/svg" version="1.1">
  <desc>A red circle with a blue border.</desc>
  <circle cx="100" cy="100" r="20"
       fill="red" stroke="blue" stroke-width="1" />
</svg>"""

display(SVG(svg))
```
