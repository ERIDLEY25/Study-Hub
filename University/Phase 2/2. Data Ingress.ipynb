## What is Data Ingress?
Data ingress means getting data into a program. Data can come from files, databases, websites, sensors, and other sources. It is the first step in a data science workflow.

```python
sources = ["file", "database", "website", "sensor"]
for s in sources:
    print(s)
```

## Input and Output (I/O)
Input = data coming into a program.

Output = data leaving a program.

Examples: reading files, displaying results, saving reports.

```python
name = input("Enter your name: ")
print("Hello", name)
```

## Working with Files
Common file types:
- TXT: plain text
- CSV: tables of data
- JSON: nested data
- XML: tagged hierarchical data

## File Modes
- r = read
- w = write (overwrite)
- a = append (add to end)

```python
with open("example.txt", "w") as f:
    f.write("Hello world")
```

## Relative vs Absolute Paths
Relative paths are preferred because they work on different computers.

Example relative path: `data/steps.csv`

## Reading Text Files Safely
Use `with open(...)` because Python closes the file automatically.

```python
with open("data/properties.txt", "r") as file:
    contents = file.read()
    print(contents)
```

## Reading Files Line by Line
Useful for large files because only one line is held in memory at a time.

```python
addresses = []
with open("data/properties.txt", "r") as file:
    for line in file:
        addresses.append(line.strip())
print(addresses)
```

## CSV Files
CSV stands for Comma-Separated Values.

Rows represent records and columns represent fields.

```python
import csv
with open("data/steps.csv", "r") as f:
    reader = csv.reader(f)
    header = next(reader)
    print(header)
    for row in reader:
        print(row)
```

## Validating Data
The staircase example checks whether measurements meet regulations.

```python
def valid_rise(rise):
    return 150 <= rise <= 220

def valid_going(going):
    return 220 <= going <= 300

print(valid_rise(150))
print(valid_rise(145))
```

## JSON Files
JSON stores data using dictionaries (objects) and lists (arrays).

Very common for web APIs and configuration files.

```python
import json
with open("data/properties.json", "r", encoding="utf-8") as f:
    data = json.load(f)
print(type(data))
```
