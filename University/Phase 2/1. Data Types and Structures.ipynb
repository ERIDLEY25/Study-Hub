# Data Types and Structures


## 1. Single Values
Scalar values store one thing at a time.

| Kind of Value | Python | Example | R | Example
| ------------- | ------ | ------- | - | -------
| Whole number | int | 2025	| integer | 2025L 
| Decimal number | float | 2.583 | numeric | 2.583
| Text | str | "Whitby"	| character	| "Whitby"
|Truth value |	bool | True	| logical | TRUE
| Nothing | None | None	| NULL | NULL



```python
year = 2025
price = 2.58
beach = "Whitby"
blue_flag = True

print(type(year))
print(type(price))
print(type(beach))
print(type(blue_flag))
```

We can change a value from one data type to another. This is called type conversion or casting.
In Python, we use the data type as a function:

```python
age = "25"      # string
age = int(age)  # convert to integer
```

When reading data from text files, everything is usually read in as a string, even if it looks like a number. We often need to convert these strings into the correct data type before using them in calculations.

#### Quick Examples

```python
num = "42"

print(int(num))    # 42
print(float(num))  # 42.0
print(str(42))     # "42"
```

#### Remember
| Conversion | Example | Result |
|------------|----------|--------|
| String → Integer | `int("42")` | `42` |
| String → Float | `float("3.14")` | `3.14` |
| Integer → String | `str(42)` | `"42"` |
| Float → Integer | `int(3.9)` | `3` |

### 1.1.1 Strings
Strings are text values.

Useful skills:
- Find length with len()
- Pick characters using indexes
- Clean text with strip()
- Split text with split()

```python
beach_name = "Whitby"
print(len(beach_name))  # the number of characters
print(beach_name[0])  # the first character
print(beach_name[-1])  # the last character
print(beach_name[0:3])  # a slice: from index 0 up to, but excluding, index 3
```

```python
region = "  East Midlands "
print(region.strip())  # remove whitespace from both ends
print(region.strip().upper())  # methods can be chained
print("Mablethorpe Beach,East Lindsey District Council".split(","))
print("Beach" in "Mablethorpe Beach")  # is one string inside another?
```

## 2. Lists
A **list** is a collection of values stored in a specific order.

In Python, lists are written using square brackets `[]`, and the items are separated by commas.

Lists are **mutable**, which means they can be changed after they are created. You can add, remove, or update items whenever needed.

```python
thanet_beaches = ["Botany Bay", "Joss Bay", "Margate Main Sands", "Minnis Bay"]
print(thanet_beaches)
print(len(thanet_beaches))  # the number of items
print(thanet_beaches[0])  # the first item
print(thanet_beaches[-1])  # the last item
print(thanet_beaches[1:3])  # a slice, which is also a list
```

Lists use **indexing** and **slicing** in the same way as strings.

The main difference is that **lists can be changed**, while strings cannot.

You can:
- Access items using their index.
- Get part of a list using slicing.
- Replace items by assigning a new value to an index.
- Add new items using `.append()`.
- Check if an item exists using `in`.

```python
thanet_beaches[2] = "Margate"  # replace the third item
print(thanet_beaches)
thanet_beaches[2] = "Margate Main Sands"  # and put it back

thanet_beaches.append("Ramsgate Main Sands")  # add to the end
thanet_beaches.append("St Mildred's Bay")
print(thanet_beaches)
print(len(thanet_beaches))
print("Joss Bay" in thanet_beaches)
print("Whitby" in thanet_beaches)
```

### 2.1 Loop Through a List
A `for` loop can go through the items of any list, one at a time, with no need to keep track of an index ourselves.

```python
for beach in thanet_beaches:
    print(f"{beach} has a Blue Flag")
```

### 2.2 Using a Loop to Summarise a List

A common pattern in programming is to use a **loop** and an **accumulator** together.

An accumulator is a variable that starts at a value (often `0`) and is updated each time the loop runs.

This allows us to process every item in a list and calculate a total, average, or other summary.

For example, we could store journey times in a list and use a loop to find the total travel time.

```python
drive_hours = [2.58, 1.62, 1.48, 2.95]

total = 0
for hours in drive_hours:
    total += hours
print("Total:", total)
print("Mean:", total / len(drive_hours))
```



#### How It Works

| Step | What Happens |
|--------|-------------|
| 1 | `total_time` starts at `0`. |
| 2 | The loop looks at each value in the list. |
| 3 | Each value is added to `total_time`. |
| 4 | After the loop finishes, `total_time` contains the final total. |

#### Remember

- A **loop** processes each item in a list.
- An **accumulator** keeps track of a running total.
- A common starting value for an accumulator is `0`.
- This pattern is useful for calculating totals, averages, and counts.

Summarising a list is so common that Python provides built-in functions for many of these jobs: `sum`, `min`, `max`, and `len`. R has `sum`, `min`, `max`, `length`, and `mean`.

##### Exercise
Write a function called `count_long_drives` that takes a list of drive times, in hours, and a number of hours, and returns how many of the drive times are longer than that number of hours. Check it with `drive_hours` and `2` hours.

```python
def count_long_drives (drive_times, max_hours):
    count = 0
    for hours in drive_times: 
        if hours > max_hours: 
            count += 1
    return count

print(count_long_drives(drive_hours, 2))
```

### 2.3. Lists of Lists
This is a simple way to store table-like data.
Each inner list represents a row.

```python
beaches = [
    ["Mablethorpe Beach", "East Lindsey District Council", "East Midlands"],
    ["Skegness Beach", "East Lindsey District Council", "East Midlands"],
    ["Whitley Bay South", "North Tyneside MBC", "North East"],
    ["Roker Beach", "Sunderland City Council", "North East"],
    ["Whitby", "North Yorkshire Council", "Yorkshire & Humber"],
]

print(beaches[2])  # the third beach: a list
print(beaches[2][0])  # the first item of the third beach: its name
```

We read `beaches[2][0]` from left to right: `beaches[2]` is the third row, and `[0]` then picks out the first item of that row.

##### Exercise
The following two-dimensional list contains animal emojis.

`animals = [["🐈", "🐆", "🐎"], ["🐓", "🐦", "🦆"], ["🦀", "🐙", "🐟"]]`

Write a function which returns a one-dimensional list containing the last item in each sub-list. The output for the above list should be `["🐎", "🦆", "🐟"]`. The function should work for any list of lists.

```python
def get_last(nested_list):  # our function takes a list of lists as input
    list_of_lasts = []  # create a new empty list to store the last elements
    for sublist in nested_list:  # loop over each sub-list in the input list
        list_of_lasts.append(sublist[-1])  # store the last element
    return list_of_lasts  # return the list of last elements


# check the function works with an example
print(get_last([[1, 2], [3, 4]]))  # should output [2, 4]

# check that the function works as expected for our case
animals = [["🐈", "🐆", "🐎"], ["🐓", "🐦", "🦆"], ["🦀", "🐙", "🐟"]]
print(get_last(animals))  # should output ['🐎', '🦆', '🐟']

# trying this for some data relating to our scenario
print(get_last(beaches))  # the region of each beach
```

## 3. Tuples
Tuples are like lists but cannot be changed after creation.
Use them for fixed records.

```python
whitby = ("Whitby", "North Yorkshire Council", "Yorkshire & Humber")

print(whitby[0])
print(len(whitby))
```

## 4. Dictionaries
Dictionaries store key-value pairs.

Think of them like a lookup table.

```python
drive_times = {
    "Whitby": 2.58,
    "Skegness": 1.62
}

print(drive_times["Whitby"])

drive_times["Roker Beach"] = 2.95
print(drive_times)
```

### Counting with Dictionaries
A very common data analysis task.

```python
regions = ["North East", "North East", "Yorkshire"]
counts = {}

for region in regions:
    counts[region] = counts.get(region, 0) + 1

print(counts)
```

## 5. Sets
Sets store unique values only.
They automatically remove duplicates.

```python
councils = ["Council A", "Council A", "Council B"]

unique_councils = set(councils)
print(unique_councils)
```

## 6. Choosing the Right Structure

| Structure | Ordered? | Changeable? | Best Use |
|-----------|----------|-------------|----------|
| List | Yes | Yes | Multiple items |
| Tuple | Yes | No | Fixed records |
| Dictionary | Key-based | Yes | Fast lookups |
| Set | No | Yes | Unique values |

## 7. Mutability

Mutable objects can change after creation.

Mutable: list, dict, set
Immutable: int, float, str, tuple, bool

```python
a = [1, 2]
b = a

b.append(3)

print(a)
print(b)
```

## 8. Comprehensions
Comprehensions create new collections in one line.

```python
numbers = [1, 2, 3, 4, 5]

squares = [n ** 2 for n in numbers]
print(squares)
```

## Quick Memory Guide

- List = ordered collection
- Tuple = fixed collection
- Dictionary = key → value lookup
- Set = unique values only
- String = text
- int/float = numbers

### Exam Tip
Ask yourself: Do I need order, lookups, or uniqueness? The answer usually tells you which structure to use.
