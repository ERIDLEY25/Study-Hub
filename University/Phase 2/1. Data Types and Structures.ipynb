## 2. Lists
A **list** is a collection of values stored in a specific order.

In Python, lists are written using square brackets `[]`, and the items are separated by commas.

Lists are **mutable**, which means they can be changed after they are created. You can add, remove, or update items whenever needed.

```python
thanet_beaches = ["Botany Bay", "Joss Bay", "Margate Main Sands", "Minnis Bay"]
print(thanet_beaches)
print(len(thanet_beaches))  # the number of items
print(thanet_beaches[0])  # the first item
print(thanet_beaches[-1])  # the last item
print(thanet_beaches[1:3])  # a slice, which is also a list
```

Lists use **indexing** and **slicing** in the same way as strings.

The main difference is that **lists can be changed**, while strings cannot.

You can:
- Access items using their index.
- Get part of a list using slicing.
- Replace items by assigning a new value to an index.
- Add new items using `.append()`.
- Check if an item exists using `in`.

```python
thanet_beaches[2] = "Margate"  # replace the third item
print(thanet_beaches)
thanet_beaches[2] = "Margate Main Sands"  # and put it back

thanet_beaches.append("Ramsgate Main Sands")  # add to the end
thanet_beaches.append("St Mildred's Bay")
print(thanet_beaches)
print(len(thanet_beaches))
print("Joss Bay" in thanet_beaches)
print("Whitby" in thanet_beaches)
```

### 2.1 Loop Through a List
A `for` loop can go through the items of any list, one at a time, with no need to keep track of an index ourselves.

```python
for beach in thanet_beaches:
    print(f"{beach} has a Blue Flag")
```

### 2.2 Using a Loop to Summarise a List

A common pattern in programming is to use a **loop** and an **accumulator** together.

An accumulator is a variable that starts at a value (often `0`) and is updated each time the loop runs.

This allows us to process every item in a list and calculate a total, average, or other summary.

For example, we could store journey times in a list and use a loop to find the total travel time.

```python
drive_hours = [2.58, 1.62, 1.48, 2.95]

total = 0
for hours in drive_hours:
    total += hours
print("Total:", total)
print("Mean:", total / len(drive_hours))
```



#### How It Works

| Step | What Happens |
|--------|-------------|
| 1 | `total_time` starts at `0`. |
| 2 | The loop looks at each value in the list. |
| 3 | Each value is added to `total_time`. |
| 4 | After the loop finishes, `total_time` contains the final total. |

#### Remember

- A **loop** processes each item in a list.
- An **accumulator** keeps track of a running total.
- A common starting value for an accumulator is `0`.
- This pattern is useful for calculating totals, averages, and counts.

Summarising a list is so common that Python provides built-in functions for many of these jobs: `sum`, `min`, `max`, and `len`. R has `sum`, `min`, `max`, `length`, and `mean`.

Write a function called `count_long_drives` that takes a list of drive times, in hours, and a number of hours, and returns how many of the drive times are longer than that number of hours. Check it with `drive_hours` and `2` hours.

### 2.3. Lists of Lists
This is a simple way to store table-like data.
Each inner list represents a row.

```python
beaches = [
    ["Mablethorpe Beach", "East Lindsey District Council", "East Midlands"],
    ["Skegness Beach", "East Lindsey District Council", "East Midlands"],
    ["Whitley Bay South", "North Tyneside MBC", "North East"],
    ["Roker Beach", "Sunderland City Council", "North East"],
    ["Whitby", "North Yorkshire Council", "Yorkshire & Humber"],
]

print(beaches[2])  # the third beach: a list
print(beaches[2][0])  # the first item of the third beach: its name
```

We read `beaches[2][0]` from left to right: `beaches[2]` is the third row, and `[0]` then picks out the first item of that row.

## 3. Tuples
Tuples are like lists but cannot be changed after creation.
Use them for fixed records. They are written with round brackets `()`

```python
whitby = ("Whitby", "North Yorkshire Council", "Yorkshire & Humber")

print(whitby[0])
print(len(whitby))
```

A particularly handy feature of tuples (and lists) is unpacking, where we assign each item to its own variable in a single line. This makes code that works with records much easier to read than code full of indices.

```python
name, council, region = whitby  # one variable per item
print(f"{name} is managed by {council}, in {region}")
```

## 4. Dictionaries
A dictionary is a collection of key-value pairs, where each unique key maps to a value. In Python, dictionaries are written in curly brackets {}, with a colon : between each key and its value. We look up a value by its key, using square brackets.

```python
drive_times = {
    "Whitby": 2.58,
    "Skegness Beach": 1.62,
    "Mablethorpe Beach": 1.48,
}
print(drive_times["Whitby"])  # look up a value by its key
drive_times["Roker Beach"] = 2.95  # add a new key-value pair
drive_times["Whitby"] = 2.6  # replace the value for an existing key
print(drive_times)
print(len(drive_times))
```

Assigning to a key works whether or not the key exists already: if it does, its value is replaced; if not, a new pair is added. Looking up a key that doesn’t exist, however, is an error. We can check first with `in`, which for a dictionary checks the keys, or use the `get` method, which gives back None (or a default of our choosing) instead of raising an error.

```python
print("Whitby" in drive_times)
print("Brighton" in drive_times)
print(drive_times.get("Brighton"))  # None, rather than an error
print(drive_times.get("Brighton", 0))  # or a default value
drive_times["Brighton"]  # a key that doesn't exist
```

Looping over a dictionary gives us its keys. More often, we want each key together with its value, which the `items` method provides as a sequence of (`key, value`) tuples that we can unpack in the loop. The `keys` and `values` methods give us just the keys or just the values.

Dictionary keys must be unique, and they must be of an unchangeable type, such as strings, numbers, or tuples (but not lists). Values can be anything at all, including lists or other dictionaries.

```python
for beach, hours in drive_times.items():
    print(f"{beach}: {hours} hours")

print(sum(drive_times.values()))
```

### Counting with Dictionaries
One of the most useful patterns in data processing is to use a dictionary to count things. We go through our data, and for each item, we add one to its count, starting the count at zero the first time that we see the item.

```python
beaches_per_region = {}
for name, council, region in beaches:
    if region not in beaches_per_region:  # first time we've seen this region
        beaches_per_region[region] = 0
    beaches_per_region[region] += 1
print(beaches_per_region)
```

The check is important. Without it, `beaches_per_region[region] += 1 `would try to look up a region that isn’t in the dictionary yet, and fail. Using get with a default of 0 is a shorter way of writing the same thing: `beaches_per_region[region] = beaches_per_region.get(region, 0) + 1.`

#### Exercise

A big part of programming often involves taking data in one data structure and transforming it into another data structure for further processing. Lists are a great way to store collections of items, but sometimes we need to convert them into dictionaries for easier access to specific elements.

Imagine we have a list of lists of beach information, where each sub-list now has a fourth item: the driving time in hours from the University of Nottingham. Write a function called beach_times_lookup that turns a list like this into a dictionary, where the beach name is the key and the driving time is the value, so that we can easily look up a beach’s driving time given its name.

```python
beach_drives = [
    ["Whitby", "North Yorkshire Council", "Yorkshire & Humber", 2.58],
    ["Skegness Beach", "East Lindsey District Council", "East Midlands", 1.62],
    ["Roker Beach", "Sunderland City Council", "North East", 2.95],
]

def beach_times_lookup(beaches, name_index=0, time_index=3):
    beach_times = {}
    for beach in beaches:
        beach_times[beach[name_index]] = beach[time_index]
    return beach_times


print(beach_times_lookup(beach_drives))
```

## 5. Sets
Sets store unique values only.
They automatically remove duplicates.

Sets have no order, so we can’t index them; `unique_councils[0] is an error`. If we need the values in order, we can use `sorted(unique_councils)`, which gives back a sorted list. Sets also support the operations from mathematical set theory, such as union (|), intersection (&), and difference (-), which we’ll return to when we look at probability later in the programme.

```python
councils = []
for name, council, region in beaches:
    councils.append(council)
print(len(councils))

unique_councils = set(councils)
print(unique_councils)
print(len(unique_councils))
```

## 8. Comprehensions
Building a new list by looping over another one is so common that Python has a compact syntax for it, called a list comprehension. The following two pieces of code do exactly the same thing.

```python
# with a loop and append
names = []
for name, council, region in beaches:
    names.append(name)
print(names)

# with a list comprehension
names = [name for name, council, region in beaches]
print(names)
```

A comprehension reads almost like English: “a list of name, for each name, council, region in beaches”. We can also add a condition to keep only some of the items, and there are equivalent comprehensions for dictionaries and sets.

```python
# only the beaches in the North East
north_east = [name for name, council, region in beaches if region == "North East"]
print(north_east)

# a dictionary of the length of each beach's name
name_lengths = {name: len(name) for name, council, region in beaches}
print(name_lengths)

# a set of the regions
regions = {region for name, council, region in beaches}
print(regions)
```
