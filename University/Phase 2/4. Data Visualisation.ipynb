## 1. Why do we visualise data?

Data visualisation helps us understand patterns, trends and relationships that are difficult to spot from numbers alone.

### Key idea
Two datasets can have similar statistics but look completely different when plotted. Always look at the data as well as the summary statistics.

## 2. Choosing the right chart

| Question | Good chart |
|---|---|
| Compare categories | Bar chart |
| Show change over time | Line chart |
| Show relationship between two numbers | Scatter plot |
| Show distribution | Histogram or Box plot |

## 3. Principles of effective visualisation

- Choose the correct chart type.
- Add titles and axis labels.
- Do not mislead the reader.
- Think about your audience.
- Make charts accessible using clear colours, labels and symbols.

## 4. Plotly Basics

Most Plotly charts follow four steps:
1. Create a figure
2. Add a trace (the data)
3. Customise the layout
4. Display the chart

```python
import plotly.graph_objects as go

x = [1,2,3,4,5]
y = [1,4,9,16,25]

fig = go.Figure(data=go.Scatter(x=x, y=y, mode="markers+lines"))
fig.update_layout(
    title="Graph of x²",
    xaxis_title="x",
    yaxis_title="y",
    template="plotly_white"
)
fig.show()
```

## 5. Understanding a Plotly Figure

### Figure
The overall chart.

### Trace
One set of data on a chart.

### Layout
Titles, labels, colours, fonts and styling.

```python
import plotly.graph_objects as go

x = [1,2,3,4,5]

fig = go.Figure(data=[
    go.Scatter(x=x, y=[1,4,9,16,25], name="x²"),
    go.Scatter(x=x, y=[1,8,27,64,125], name="x³")
])

fig.update_layout(title="Multiple Traces")
fig.show()
```

## 7. Accessibility Tips

Remember:
- Use readable font sizes.
- Use colour-blind friendly colours.
- Do not rely only on colour.
- Use different marker shapes and line styles.
- Make sure charts still work in black and white.

```python
import plotly.graph_objects as go

x=[1,2,3,4,5]

fig=go.Figure()
fig.add_trace(go.Scatter(
    x=x,y=[1,4,9,16,25],
    mode="markers+lines",
    marker=dict(color="#E69F00"),
    name="x²"
))
fig.update_layout(font=dict(size=14))
fig.show()
```

## 9. Plotly Express

Plotly Express is a simpler way to create charts.

You provide a DataFrame and tell Plotly which columns to use.

```python
import pandas as pd
import plotly.express as px

data = pd.DataFrame({
    "Snack": ["Tea", "Biscuits", "Crisps", "Cake"],
    "Votes": [40,30,20,10]
})

fig = px.pie(data, names="Snack", values="Votes", title="Office Snack Poll")
fig.show()
```
