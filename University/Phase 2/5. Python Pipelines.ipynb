# Python Pipelines

A simple notebook to help you remember and understand Python pipelines.

## What is a Pipeline?

A pipeline is a series of steps where the output from one step becomes the input for the next step.

Typical stages:
1. Source - get the data
2. Load - read the data into Python
3. Process - clean or transform the data
4. Output - show results in a chart, report, or file

````python
# A simple pipeline example
numbers = [1, 2, 3, 4, 5]
doubled = [n * 2 for n in numbers]
total = sum(doubled)
print(total)
````

## Why Use Pipelines?

- Easier to read
- Easier to test
- Functions can be reused
- Easier to maintain

## One Function Per Step

A good pipeline usually has one function for each task.

````python
def load_data():
    return [1,2,3,4]

def process_data(data):
    return [x * 10 for x in data]

def output_data(data):
    print(data)

data = load_data()
data = process_data(data)
output_data(data)
````

## Airline Example Overview

The course example uses two CSV files:

- Customer flights
- Airport distances

The pipeline: Load → Create lookup → Merge data → Calculate totals → Sort → Create chart

````python
def generate_lookup(distances):
    lookup = {}
    for airport_from, airport_to, distance in distances:
        lookup[(airport_from, airport_to)] = distance
    return lookup
````

## Key Idea: Dictionaries for Fast Lookups

Instead of searching a whole list every time, store values in a dictionary and access them quickly using a key.

## Pipeline Structure

```text
CSV Files
    ↓
Read Files
    ↓
Create Lookup Dictionary
    ↓
Merge Data
    ↓
Calculate Totals
    ↓
Sort Results
    ↓
Create Chart
```

## Handling Errors

Use try/except so one bad file does not stop the whole pipeline.

````python
try:
    number = int("abc")
except ValueError:
    print("Could not convert to a number")
````
